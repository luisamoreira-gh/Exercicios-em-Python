```python
#6. Imagine que você está desenvolvendo um navegador web simplificado. Use uma pilha para armazenar o histórico de páginas visitadas pelos usuários e implementar as funcionalidades de voltar e avançar na
# navegação.
class Pilha:
  def __init__(self):
    self.items = []
  def empilhar(self, item):
    self.items = self.items + [item]
  def desempilhar(self):
    if not self.is_vazia():
      item_removido = self.items[-1]
      self.items = self.items[:-1]
      return item_removido
    else:
      print("A pilha está vazia. Não é possível desempilhar.")
  def topo(self):
    if not self.is_vazia():
      return self.items[-1]
    else:
      print("A pilha está vazia. Não há topo para visualizar.")
  def is_vazia(self):
    return len(self.items) == 0
  def tamanho(self):
    return len(self.items)
  def imprimir(self):
    if not self.is_vazia():
      print(f"Itens da pilha: {self.items}")
    else:
      print("A pilha está vazia. Não há itens para imprimir.")
  def reverse(self):
    return self.items[::-1]

paginas = Pilha()

#Empilhar = adiciona uma nova página à pilha = avança na navegação
paginas.empilhar(1)
paginas.empilhar(2)
paginas.empilhar(3)
paginas.empilhar(4)
paginas.empilhar(5)
paginas.empilhar(6)

paginas.imprimir()

#Desempilhar = remove a última pagina adicionada = volta na página anterior de navegação
paginas.desempilhar()
paginas.desempilhar()

paginas.imprimir()
```

```python
#7. Crie uma calculadora que avalia expressões matemáticas no formato Notação Polonesa Reversa (RPN). Use uma pilha para armazenar os operandos e operadores e realizar os cálculos.
def RPN(equacao, pilha):
  elementos = equacao.split()
  for i in elementos:
    if i not in ("+", "-", "*", "/", "^"):
      pilha.empilhar(float(i))
    else:
      b = pilha.topo()
      pilha.desempilhar()
      a = pilha.topo()
      pilha.desempilhar()
      if i == "+":
        calculo = a + b
      elif i == "-":
        calculo = a - b
      elif i == "*":
        calculo = a * b
      elif i == "/":
        calculo = a / b
      elif i == "^":
        calculo = a ** b
      pilha.empilhar(calculo)
  return f"O resultado da equação '{equacao}' é = {pilha.topo()}"

pilha = Pilha()
equacao = "5 4 * 2 3 * - 2 1 * /"
print(RPN(equacao, pilha))
```

```python
#8. Em um programa de edição de texto, implemente a funcionalidade de "Desfazer" e "Refazer" usando uma pilha para armazenar o histórico de comandos executados pelo usuário.
comandos = Pilha()
deletado = Pilha()

#"Desfazer" = Desempilhar + guardar a informação anterior
def desfazer(lista, deletado):
  if lista.tamanho() != 0:
    item = lista.desempilhar()
    deletado.empilhar(item)
    return lista
  else:
    print("Não há nenhum comando prévio.")

#"Refazer" = Empilhar a informação anterior
def refazer(lista, deletado):
  if deletado.tamanho() != 0:
    item = deletado.desempilhar()
    lista.empilhar(item)
    return lista
  else:
    print("Não há nenhum comando para refazer")

comandos.empilhar(1)
comandos.empilhar(2)
comandos.empilhar(3)

comandos.imprimir()
desfazer(comandos, deletado)
comandos.imprimir()
refazer(comandos, deletado)
comandos.imprimir()
```
