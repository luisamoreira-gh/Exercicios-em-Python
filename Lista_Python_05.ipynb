```python
#1. Crie uma classe chamada “Circulo” que tenha um atributo “raio”. Implemente um método chamado “calcular_area” que retorna a área do círculo.
class Circulo:
  def __init__(self, raio):
    self.raio = raio
  def calcular_area(self):
    self.area = 3.14 * (self.raio ** 2)
    print(f"Área do círculo de raio {self.raio}: {self.area}")

valor_raio = Circulo(float(input("Digite o valor do raio desejado: ")))
Circulo.calcular_area(valor_raio)
```

```python
#3. Crie uma classe chamada “Retangulo” que tenha atributos “base” e “altura”. Implemente um método chamado “calcular_area” que retorna a área do retângulo.
class Retangulo:
  def __init__(self, base, altura):
    self.base = base
    self.altura = altura
  def calcular_area(self):
    self.area = self.base * self.altura
    print(f"Área do retângulo de lados {self.base} e {self.altura}: {self.area}")

valor_base = Retangulo(2.0, 4.0)
Retangulo.calcular_area(valor_base)
```

```python
#4. Crie uma classe chamada “ContaBancaria” que tenha atributos “saldo” e “titular”. Implemente métodos “depositar” e “sacar” para manipular o saldo.
class ContaBancaria:
  def __init__(self, saldo, titular):
    self.saldo = saldo
    self.titular = titular
  def deposito(self, valor):
    self.valor = valor
    self.deposito = self.saldo + self.valor
    print(f"Seu saldo é {self.deposito}")
  def saldo(self, valor):
    self.valor = valor
    if self.valor <= self.saldo:
      self.sacar = self.saldo - self.valor
      print(f"Seu saldo é {self.sacar}")
    else:
      print(f"Saldo insuficiente.")

valor_base = ContaBancaria(200, "Amanda")
ContaBancaria.deposito(valor_base, 200)
ContaBancaria.saldo(valor_base, 200)
ContaBancaria.saldo(valor_base, 500)
```

```python
#6. Crie uma classe chamada “Produto” com atributos “nome”, “preco” e “quantidade”. Implemente um método chamado “calcular_total” que retorna o valor total do produto (preço * quantidade).
class Produto:
  def __init__(self, nome, preco, quantidade):
    self.nome = nome
    self.preco = preco
    self.quantidade = quantidade
  def calcular_total(self):
    self.valor = self.preco * self.quantidade
    self.valor = round(self.valor, 2)
    print(f"{self.nome} ({self.quantidade} unid. x R${self.preco}): R${self.valor}")

item = Produto("Detergente", 3.99, 11)
item.calcular_total()
```

```python
#7. Crie uma classe chamada “Carro” com atributos “marca”, “modelo” e “ano”. Implemente um método chamado “detalhes” que retorna uma string com as informações do carro.
class Carro:
  def __init__(self, marca, modelo, ano):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano
  def detalhes(self):
    print(f"Detalhes do carro:")
    print(f"Marca = {self.marca}; Modelo = {self.modelo}; Ano = {self.ano}.")

carro = Carro("Fiat", "Mobi", 2016)
carro.detalhes()
```

```python
#8. Crie uma classe chamada “Aluno” com atributos “nome” e “notas”. Implemente um método chamado “calcular_media” que retorna a média das notas do aluno.
class Aluno:
  def __init__(self, nome, notas):
    self.nome = nome
    self.notas = notas
  def calcular_media(self):
    self.media = sum(self.notas) / len(self.notas)
    self.media = round(self.media, 1)
    if self.media < 7:
      print(f"Aluno(a): {self.nome}; Média geral: {self.media}. Aluno reprovado.")
    else:
      print(f"Aluno(a): {self.nome}; Média geral: {self.media}. Aluno aprovado.")

aluno1 = Aluno("Carlos", [9.0, 1.5, 6.8])
aluno2 = Aluno("Joana", [9.8, 10.0, 5.7])
aluno1.calcular_media()
aluno2.calcular_media()
```

```python
#9. Crie uma classe chamada “Triangulo” com atributos “lado1”, “lado2” e “lado3”. Implemente um método chamado “calcular_perimetro” que retorna o perímetro do triângulo.
class Triangulo:
  def __init__(self, lado1, lado2, lado3):
    self.lado1 = lado1
    self.lado2 = lado2
    self.lado3 = lado3
  def calcular_perimetro(self):
    self.perimetro = self.lado1 + self.lado2 + self.lado3
    print(f"Perímetro = {self.perimetro}cm")

triangulo = Triangulo(2, 3, 5)
triangulo.calcular_perimetro()
```

```python
#10. Crie uma classe chamada “Funcionario” com atributos “nome”, “salario” e “cargo”. Implemente um método chamado “aumentar_salario” que recebe um valor percentual de aumento e atualiza o salário do funcionário.
class Funcionario:
  def __init__(self, nome, salario, cargo):
    self.nome = nome
    self.salario = salario
    self.cargo = cargo
  def aumentar_salario(self):
    self.novosalario = self.salario * 1.35
    self.novosalario = round(self.novosalario)
    print(f"O salário do(a) funcionário(a) {self.nome}, no cargo de {self.cargo}, recebeu um aumento. Seu novo salário é R${self.novosalario}")

pessoa = Funcionario("Alexa", 2379, "Secretário(a)")
pessoa.aumentar_salario()
```
