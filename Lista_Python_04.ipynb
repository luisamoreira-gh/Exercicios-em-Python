```python
# 1. Faça um programa que calcule a média de três números inseridos pelo usuário. (Resolvido de duas maneiras)
def media1():
  a = int(input(print("Digite o 1- número: ")))
  b = int(input(print("Digite o 2- número: ")))
  c = int(input(print("Digite o 3- número: ")))
  media = (a + b + c) / 3
  print(f"Sua média é {media}")
def media2():
  soma = 0
  for i in range(3):
    a = int(input(print("Digite seu número: ")))
    soma += a
  print(f"Sua média é {soma / 3}")
media1()
media2()
```

```python
# 2. Crie um programa que determine se um número inserido pelo usuário é par ou ímpar.
a = int(input(print("Digite o número: ")))
if a % 2 == 0:
  print(f"O número {a} é par.")
else:
  print(f"O número {a} é ímpar.")
```

```python
# 3. Escreva um programa que solicite um número ao usuário e imprima todos os números pares de 0 até esse número.
a = int(input(print("Digite o número: ")))
for i in range(0, a + 1, 2):
  print(i)
```

```python
# 4. Crie um programa que leia uma lista de números e exiba o maior e o menor valor da lista.
num = [12, 46, 44, 82, 23, 31, 12]
print(f"O maior número é {max(num)}")
print(f"O menor número é {min(num)}")
```

```python
# 5. Faça um programa que leia uma lista de números e retorne a média dos números pares.
num = [12, 46, 44, 82, 23, 31, 12]
pares = []
for i in range(len(num)):
  if num[i] % 2 == 0:
    pares.append(num[i])
  else:
    continue
print(f"A média dos números pares é {sum(pares) / len(pares)}")
```

```python
# 6. Escreva um programa que peça um número inteiro positivo ao usuário e calcule o fatorial desse número.
a = int(input(print("Digite o número: ")))
original = a
fatorial = 1
while a > 1:
  fatorial *= a
  a -= 1
print(f"O fatorial de {original} é igual a {fatorial}")
```

```python
# 7. Crie um programa que imprima a sequência de Fibonacci até um valor inserido pelo usuário.
x = 0
x1 = 0
x2 = 1
a = int(input("Digite o número: "))
print(0)
while x <= a:
    x1 = x
    x = x + x2
    x2 = x1
    if x <= a:
        print(x)
    else:
        break
```

```python
# 8. Faça um programa que determine se um número é primo ou não.
a = 23
primo = True
for i in range(2, a):
  if a % i == 0:
    primo = False
    break
if primo == True:
  print(f"O número {a} é primo")
else:
  print(f"O número {a} não é primo")
```

```python
# 9. Escreva um programa que leia uma lista de nomes e retorne uma nova lista com apenas os nomes que começam com a letra 'A'.
nomes = ["Arlindo", "Bianca", "Carlos", "Amanda", "Pablo"]
A = []
for nome in nomes:
  if nome.startswith(("A", "a")):
    A.append(nome)
print(A)
```
