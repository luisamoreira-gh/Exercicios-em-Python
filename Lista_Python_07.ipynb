```python
#1. Escreva uma função em Python para ordenar um vetor de inteiros em ordem crescente usando o algoritmo de Select Sort.
def select_sort(list):
  for i in range(len(list)):
    menor = i
    for j in range(i + 1, len(list)):
      if list[menor] > list[j]:
        menor = j

    aux = list[i]
    list[i] = list[menor]
    list[menor] = aux
  return list

lista1 = [8, 2, 3, 6, 5, 1, 7, 9, 4]
print(select_sort(lista1))
```

```python
#2. Escreva uma função em Python para ordenar um vetor de inteiros, ele deve receber um parâmetro que serve como chave para realizar a ordenação crescente ou decrescente.
def ordem(list, chave):
  if chave.lower() == "crescente":
    for i in range(len(list)):
      menor = i
      for j in range(i + 1, len(list)):
        if list[menor] > list[j]:
          menor = j
      aux = list[i]
      list[i] = list[menor]
      list[menor] = aux
    return list

  elif chave.lower() == "decrescente":
    for i in range(len(list)):
      maior = i
      for j in range(i + 1, len(list)):
        if list[maior] < list[j]:
          maior = j
      aux = list[i]
      list[i] = list[maior]
      list[maior] = aux
    return list
  else:
    return "Desculpe, esse comando não existe."

lista2 = [8, 2, 3, 6, 5, 1, 7, 9, 4]
print(ordem(lista2, "Crescente"))
print(ordem(lista2, "Decrescente"))
print(ordem(lista2, "abobrinha"))
```

```python
#3. Escreva um programa que encontre o elemento de maior valor em um vetor de inteiros não ordenado sem usar a função `max()`. Em seguida, encontre o elemento mínimo sem usar a função `min()`.
def min_max(list):
  menor = 0
  maior = 0
  for j in range(1, len(list)):
    if list[menor] > list[j]:
      menor = j
    if list[maior] < list[j]:
      maior = j
  return f"O menor número da lista é {list[menor]}. O maior número da lista é {list[maior]}."

lista3 = [8, 2, 3, 6, 5, 1, 7, 9, 4]
print(min_max(lista3))
```

```python
#5. Implemente uma função que aceite um vetor de números inteiros e remova todos os elementos duplicados, retornando o vetor resultante sem duplicatas. Não é permitido utilizar a função “set()”
def sai_dupe(list):
  lista_limpa = []
  for item in list:
    if item not in lista_limpa:
      lista_limpa.append(item)
  return lista_limpa

lista5 = [5, 4, 5, 1, 2, 2, 3]
print(sai_dupe(lista5))
```
